```python
# Splitting the dataset into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42)
```

X_train = X_train[top_10_features]

```python
# Check if X_train and X_val are Pandas DataFrames, convert them to NumPy arrays if necessary
X_train_np = X_train.to_numpy() if isinstance(X_train, pd.DataFrame) else X_train
X_val_np = X_val.to_numpy() if isinstance(X_val, pd.DataFrame) else X_val

# Ensure that all features are numeric
print(X_train_np.dtype)
print(X_val_np.dtype)


# Convert to DMatrix for XGBoost
train_dmatrix = xgb.DMatrix(X_train_np, label=y_train)
val_dmatrix = xgb.DMatrix(X_val_np, label=y_val)
```

```python

from sklearn.model_selection import GridSearchCV

param_grid = {
   
    'max_depth': [7],
    'learning_rate': [0.01],
    'subsample': [1.0],
    'colsample_bytree': [0.8],
    'reg_alpha': [0.1],  # L1 regularization
    'reg_lambda': [100.0],  # L2 regularization
    'min_child_weight': 5,  # Minimum sum of instance weight in a leaf
    'gamma': 0.5  
}

# Pass the params to GridSearchCV
grid_search = GridSearchCV(estimator=xgb.XGBRegressor(), param_grid=param_grid, scoring='neg_mean_squared_error', cv=3, n_jobs=-1, verbose=2)

# Train the model using the best combination of parameters
grid_search.fit(X_train, y_train)

# Retrieve the best parameters found by GridSearchCV
print(grid_search.best_params_)
```

```python
# Set the hyperparameters for XGBoost
params = {

    'max_depth': 12,
    'learning_rate': 0.01,
    'subsample': 1.0,
    'colsample_bytree': 0.8,
    'reg_alpha': 0.1,  # L1 regularization
    'reg_lambda': 200,  # L2 regularization
    'min_child_weight': 5,  # Minimum sum of instance weight in a leaf
    'gamma': 0.5                      # L1 regularization term           
}
```

```python

# Train the model using XGBoost's train function
xgb_model = xgb.train(params, train_dmatrix, num_boost_round=100000)
```

```python
## Convert X_train and X_val to DMatrix
from sklearn.metrics import mean_squared_error  # Import this to calculate RMSE
# Check if X_train and X_val are Pandas DataFrames, convert them to NumPy arrays if necessary
X_train_np = X_train.to_numpy() if isinstance(X_train, pd.DataFrame) else X_train
X_val_np = X_val.to_numpy() if isinstance(X_val, pd.DataFrame) else X_val

# Ensure that all f


train_dmatrix = xgb.DMatrix(X_train_np, label=y_train)
val_dmatrix = xgb.DMatrix(X_val_np, label=y_val)

# Predict on the training set
y_train_pred = xgb_model.predict(train_dmatrix)

# Predict on the validation set
y_val_pred = xgb_model.predict(val_dmatrix)

# Calculate RMSE on the training set
train_rmse = mean_squared_error(y_train, y_train_pred, squared=False)
print(f"Training RMSE: {train_rmse}")

# Calculate RMSE on the validation set
val_rmse = mean_squared_error(y_val, y_val_pred, squared=False)
print(f"Validation RMSE: {val_rmse}")
```

```python
import numpy as np

# Mean Directional Accuracy (MDA)
def mean_directional_accuracy(y_true, y_pred):
    # Ensure y_true is a NumPy array
    if isinstance(y_true, pd.DataFrame) or isinstance(y_true, pd.Series):
        y_true = y_true.values  # Convert to NumPy array if it's a DataFrame or Series
    
    # Ensure both arrays have the same length after shifting
    y_true_diff = np.sign(y_true[1:] - y_true[:-1])
    y_pred_diff = np.sign(y_pred[1:] - y_pred[:-1])
    return np.mean(y_true_diff == y_pred_diff)

# Convert y_val to NumPy array if it is a DataFrame or Series
y_val_np = y_val.values if isinstance(y_val, pd.DataFrame) else y_val

# Calculate MDA
mda = mean_directional_accuracy(y_val_np, y_val_pred)
print(f"Mean Directional Accuracy: {mda}")
```
